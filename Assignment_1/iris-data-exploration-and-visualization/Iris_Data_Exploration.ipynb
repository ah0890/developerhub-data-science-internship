# 🌸 Iris Dataset — Exploratory Data Analysis & Visualization

**Objective:** Perform comprehensive EDA on the classic Iris dataset and uncover patterns that distinguish the three flower species.

**Dataset:** Iris — 150 records · 4 features · 3 species  
**Source:** `sklearn.datasets.load_iris()` (built-in)

---
### Project Workflow
1. Import Libraries  
2. Load & Inspect Dataset  
3. Statistical Summary  
4. Univariate Analysis — Histograms, Box Plots, Violin Plots  
5. Bivariate Analysis — Scatter Plots, Pairplot  
6. Correlation Heatmap  
7. Swarm Plots  
8. Key Insights & Conclusions

## 1. Import Libraries

```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

matplotlib.rcParams['figure.dpi']     = 110
matplotlib.rcParams['figure.figsize'] = (10, 5)
sns.set_theme(style='whitegrid', palette='Set2')

SPECIES_COLORS = {'setosa': '#e74c3c', 'versicolor': '#3498db', 'virginica': '#2ecc71'}

print('✅  Libraries imported successfully!')
print(f'    pandas {pd.__version__}  |  numpy {np.__version__}')
```

## 2. Load & Inspect Dataset

```python
iris = load_iris()
df   = pd.DataFrame(iris.data, columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'])
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

print(f'Dataset shape : {df.shape[0]} rows x {df.shape[1]} columns')
print(f'Species       : {df.species.unique().tolist()}')
df.head(10)
```

```python
print('Column data types:')
print(df.dtypes)
print(f'\nMissing values: {df.isnull().sum().sum()}')
```

## 3. Statistical Summary

```python
df.describe().round(2)
```

```python
print('Mean values by species:')
df.groupby('species').mean().round(2)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
counts = df['species'].value_counts().sort_index()
colors = list(SPECIES_COLORS.values())

axes[0].bar(counts.index, counts.values, color=colors, edgecolor='white', linewidth=1.5)
axes[0].set_title('Species Count', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Count')
for i, v in enumerate(counts.values):
    axes[0].text(i, v + 0.5, str(v), ha='center', fontweight='bold')

axes[1].pie(counts.values, labels=counts.index, colors=colors,
            autopct='%1.1f%%', startangle=120,
            wedgeprops={'edgecolor': 'white', 'linewidth': 2})
axes[1].set_title('Species Distribution', fontsize=13, fontweight='bold')

plt.suptitle('Class Distribution — Perfectly Balanced (50 per species)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 4. Univariate Analysis

### 4.1 Histograms — Feature Distributions

```python
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
labels   = ['Sepal Length (cm)', 'Sepal Width (cm)', 'Petal Length (cm)', 'Petal Width (cm)']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for ax, feat, label in zip(axes, features, labels):
    for species, color in SPECIES_COLORS.items():
        subset = df[df['species'] == species][feat]
        ax.hist(subset, bins=15, alpha=0.65, color=color, edgecolor='white', label=species)
    ax.axvline(df[feat].mean(), color='black', linestyle='--', linewidth=1.5,
               label=f'Mean = {df[feat].mean():.2f}')
    ax.set_title(label, fontsize=12, fontweight='bold')
    ax.set_xlabel(label)
    ax.set_ylabel('Frequency')
    ax.legend(fontsize=8)

plt.suptitle('Feature Distributions by Species', fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

### 4.2 Box Plots — Spread & Outliers

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for ax, feat, label in zip(axes, features, labels):
    sns.boxplot(data=df, x='species', y=feat, palette=SPECIES_COLORS, ax=ax,
                flierprops=dict(marker='o', markerfacecolor='red', markersize=5))
    ax.set_title(label, fontsize=12, fontweight='bold')
    ax.set_xlabel('Species')
    ax.set_ylabel(label)

plt.suptitle('Box Plots — Feature Spread by Species', fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

### 4.3 Violin Plots — Distribution Shape

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for ax, feat, label in zip(axes, features, labels):
    sns.violinplot(data=df, x='species', y=feat, palette=SPECIES_COLORS, ax=ax,
                   inner='box', linewidth=1.2)
    ax.set_title(label, fontsize=12, fontweight='bold')
    ax.set_xlabel('Species')
    ax.set_ylabel(label)

plt.suptitle('Violin Plots — Distribution Shape by Species', fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

## 5. Bivariate & Multivariate Analysis

### 5.1 Scatter Plots — Key Feature Pairs

```python
pairs       = [('sepal_length','sepal_width'), ('petal_length','petal_width'),
               ('sepal_length','petal_length'), ('sepal_width','petal_width')]
pair_labels = [('Sepal Length','Sepal Width'), ('Petal Length','Petal Width'),
               ('Sepal Length','Petal Length'), ('Sepal Width','Petal Width')]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for ax, (fx, fy), (lx, ly) in zip(axes, pairs, pair_labels):
    for species, color in SPECIES_COLORS.items():
        s = df[df['species'] == species]
        ax.scatter(s[fx], s[fy], c=color, alpha=0.75, edgecolors='white',
                   linewidth=0.5, s=65, label=species)
    ax.set_xlabel(f'{lx} (cm)', fontsize=10)
    ax.set_ylabel(f'{ly} (cm)', fontsize=10)
    ax.set_title(f'{lx} vs {ly}', fontsize=11, fontweight='bold')
    ax.legend(fontsize=8)

plt.suptitle('Scatter Plots — Feature Pair Relationships', fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

### 5.2 Pairplot — All Feature Combinations

```python
g = sns.pairplot(df, hue='species', palette=SPECIES_COLORS,
                 diag_kind='kde', plot_kws={'alpha': 0.6, 'edgecolor': 'white'},
                 height=2.5)
g.figure.suptitle('Pairplot — All Feature Combinations by Species',
                  fontsize=14, fontweight='bold', y=1.02)
plt.show()
```

## 6. Correlation Heatmap

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Numerical features only
corr_all = df[features].corr()
sns.heatmap(corr_all, annot=True, fmt='.3f', cmap='RdYlGn',
            linewidths=0.5, ax=axes[0], vmin=-1, vmax=1,
            annot_kws={'size': 12, 'fontweight': 'bold'})
axes[0].set_title('Feature-to-Feature Correlation', fontsize=13, fontweight='bold')

# Include encoded species
df_enc = df.copy()
df_enc['species_code'] = df_enc['species'].cat.codes
corr_target = df_enc.corr()
sns.heatmap(corr_target, annot=True, fmt='.3f', cmap='RdYlGn',
            linewidths=0.5, ax=axes[1], vmin=-1, vmax=1,
            annot_kws={'size': 12, 'fontweight': 'bold'})
axes[1].set_title('Correlation Including Target (Species)', fontsize=13, fontweight='bold')

plt.suptitle('Correlation Analysis', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()

print('\nCorrelation with species (encoded — higher = better predictor):')
corr_with_species = df_enc.corr()['species_code'].drop('species_code').sort_values(ascending=False)
for feat, r in corr_with_species.items():
    print(f'  {feat:<15}  r = {r:+.3f}')
```

## 7. Swarm Plots — Individual Data Points

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for ax, feat, label in zip(axes, features, labels):
    sns.swarmplot(data=df, x='species', y=feat, palette=SPECIES_COLORS,
                  ax=ax, size=5, alpha=0.85)
    ax.set_title(label, fontsize=12, fontweight='bold')
    ax.set_xlabel('Species')
    ax.set_ylabel(label)

plt.suptitle('Swarm Plots — Every Individual Sample', fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

## 8. Key Insights & Conclusions

### 🔍 Feature Discriminability

| Feature | Discriminability | Key Finding |
|---------|-----------------|-------------|
| `petal_length` | ✅ **Excellent** | Setosa (1–2 cm) fully separates from the others |
| `petal_width`  | ✅ **Excellent** | Same separation pattern; r = 0.96 with petal_length |
| `sepal_length` | ⚠️ **Partial**   | Versicolor & Virginica overlap significantly |
| `sepal_width`  | ❌ **Weak**       | Heavy overlap across all three species |

### 💡 Top 5 Findings
1. **Setosa is perfectly linearly separable** from the other two species using petal dimensions alone
2. **Petal length and petal width are highly correlated** (r ≈ 0.96) — they capture the same underlying variance
3. **Versicolor and Virginica overlap** in every feature; classification needs all 4 features together
4. **Sepal features are less predictive** — sepal width is nearly useless for distinguishing species
5. **Dataset is perfectly balanced** (50 samples / species) — ideal for unbiased ML training
